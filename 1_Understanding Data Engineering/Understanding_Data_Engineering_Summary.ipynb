## Chapter 1 — What Is Data Engineering?

**The core job:** a data engineer's mandate is to deliver *the correct data*, *in the right form*, *to the right people*, *as efficiently as possible*. Concretely, that means:
- Ingesting data from many different, often inconsistent sources.
- Optimizing databases so downstream analysis is fast and reliable.
- Removing corrupted or malformed data before it reaches anyone else.
- Designing, building, testing, and maintaining the data architecture itself.

**Data engineers vs. data scientists — how they divide the work:**

| | Data Engineer | Data Scientist |
|---|---|---|
| Role | Ingests and stores data, sets up databases, builds pipelines | Exploits data, accesses databases, uses pipeline outputs |
| Core skill | Strong software engineering | Strong statistical/analytical skills |

The key insight here is that these two roles are **complementary, not overlapping**: a data engineer's pipeline is only valuable insofar as it reliably feeds a data scientist's (or analyst's) work downstream. This is the same relationship I experienced hands-on later in this specialization — the Bronze → Silver → Gold work in the Databricks SQL module *is* data engineering work, done specifically so that clean, trustworthy Gold tables are ready for analysis.

**Data pipelines and ETL:** a data pipeline automates the flow of data from one system to the next, reducing both human intervention and the errors that come with manual handling. **ETL (Extract, Transform, Load)** is the classic pattern for this: extract raw data, transform it into a usable shape, and load it into its destination. Not every pipeline strictly follows ETL — sometimes data is loaded without transformation, straight into an application — but ETL is the mental model worth defaulting to.

**Why this matters at scale — the five Vs of big data:**
- **Volume** — how much data there is.
- **Variety** — what kinds/formats of data (structured, semi-structured, unstructured).
- **Velocity** — how frequently new data arrives.
- **Veracity** — how accurate/trustworthy the data is.
- **Value** — how useful the data actually is once processed.

As data volume grows (driven by sensors/IoT, social media, enterprise systems, and streaming/VoIP traffic), traditional single-machine methods stop working, and this is precisely why data engineering exists as a distinct discipline rather than something a data scientist just does on the side.

## Chapter 2 — How Data Storage Works

**Structured, semi-structured, and unstructured data — and why the distinction matters:**

- **Structured data** — consistent rows/columns, defined types, easy to search and group. Lives in relational databases and is queried with SQL. Only about 20% of real-world data is actually structured.
- **Semi-structured data** — a consistent-ish model but with flexible, variable-size records (e.g. JSON, XML, YAML). Common in NoSQL databases. More work to organize than structured data, but still tractable.
- **Unstructured data** — no fixed model at all (text, audio, images, video). Hard to search or organize directly, usually lives in data lakes, but it's the *majority* of data generated today and can be extremely valuable once it's given some structure (e.g. via AI-based extraction).

**SQL as the industry standard:** SQL (Structured Query Language) is how relational databases are both built and queried. The split in usage mirrors the engineer/scientist divide above:
- Data **engineers** use SQL to `CREATE`, alter, and maintain tables (schema design).
- Data **scientists/analysts** use SQL to `SELECT`, filter, group, and aggregate — i.e. to ask questions of the data.

A **database schema** governs how tables relate to one another — the blueprint that keeps a multi-table database coherent rather than a pile of disconnected tables.

**Data warehouses vs. data lakes — the trade-off that shapes architecture decisions:**

| | Data Lake | Data Warehouse |
|---|---|---|
| Stores | All raw data, any structure | Curated, mostly structured data for a specific use |
| Scale | Can reach petabyte scale | Relatively small |
| Cost | Cost-effective | More costly to maintain/update |
| Analysis | Harder to analyze directly | Optimized for analysis |
| Users | Primarily data scientists (big data, real-time) | Data/business analysts (ad-hoc, read-only queries) |

A **data catalog** — documenting a dataset's source, usage, owner, and update frequency — is what keeps a data lake from degrading into a "data swamp": without governance, "store everything" quickly becomes "find nothing." This is the exact motivation behind the medallion architecture (Bronze/Silver/Gold) and Unity Catalog governance I later practiced hands-on in the Databricks SQL module — the **Lakehouse** concept is essentially "take the flexibility and low cost of a data lake, and layer in the structure, governance, and query performance of a data warehouse on top."

Finally: a **database** is the general term for organized, computer-accessible data — a data warehouse is simply one *type* of database, optimized for a specific analytical workload.
